```python
bucket_name = 'apeter40-hw6' # the destination bucket name
local_path = '/content/data' # the local notebook path of the data files
bucket_folder = 'data' # the folder name in the bucket
s3 = boto3.resource('s3')
my_bucket = s3.Bucket(bucket_name)

print ('S3 bucket : ' + my_bucket.name) # print the bucket name
for path, subdirs, files in os.walk(local_path) : # iterate through all folders and files
  path = path.replace ("\\" ,"/")
  directory_name = path.replace (local_path , "")
  for file in files:
    # print the source file information and target file information
    print (os.path.join(path , file) +  '--> ' + bucket_folder + '/' + directory_name + file)
    # upload the source file to the target file in the bucket
    my_bucket.upload_file(os.path.join(path, file), bucket_folder + '/' + directory_name + file)
```

```python
# print the name of the bucket
print ( 'S3 bucket : ' + my_bucket . name )

# sum_bytes will be used to sum up all the individual file sizes
sum_bytes = 0
# loop over all objects in my bucket
for my_bucket_object in my_bucket . objects . all () :
  print ( my_bucket_object . key ) # print the object name
  sum_bytes = sum_bytes + my_bucket_object . size # add the file size to sum_bytes

  # convert the sum of bytes to megabytes ( MB ) unit
sum_mb = sum_bytes * 0.000001
# print the totoal size in MB
print ('Total size = ' , str (sum_mb) + 'MB')
```

```python
import math
import codecs

bucket_name = 'apeter40-hw6'
bucket_folder = 'data/'

# set the S3 bucket name and folder information
s3 = boto3.resource ('s3')
my_bucket = s3.Bucket(bucket_name)

# iterate over all objects in a bucket folder
mu = 0
SD = 0
k_total = 1
for obj in my_bucket.objects.filter(Prefix=bucket_folder):
  # check if the file has the .dat extension
  if obj.key.endswith('.dat'):
    print(obj.key)
  # Create the python script to calculate the average and standard devaiation.
    with codecs.open(obj.key, 'r', encoding='utf-8', errors='ignore') as file:
      for line in file:
        mu_k_minus_1 = mu
        mu = mu + ((float(line) - mu)/k_total)
        k_total = k_total + 1
        SD = SD + ((float(line) - mu_k_minus_1)*(float(line) - mu))
# print the results
SD = math.sqrt(SD/k_total)
print('Average = ', mu) # final average value
print('Standard deviation = ' , SD) # final standard dev.value
```
